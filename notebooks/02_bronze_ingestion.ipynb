```python
from pyspark.sql.types import StructType, StructField, StringType

orders_schema = StructType([
    StructField("order_id", StringType(), True),
    StructField("customer_id", StringType(), True),
    StructField("order_status", StringType(), True),
    StructField("order_purchase_timestamp", StringType(), True),
    StructField("order_approved_at", StringType(), True),
    StructField("order_delivered_carrier_date", StringType(), True),
    StructField("order_delivered_customer_date", StringType(), True),
    StructField("order_estimated_delivery_date", StringType(), True)
])
```

```python
from pyspark.sql import functions as F
from uuid import uuid4

source_path = "/Volumes/lakehouse/raw/olist_source"
source_file = f"{source_path}/olist_orders_dataset.csv"

batch_id = str(uuid4())

orders_bronze_df = (
    spark.read
    .format("csv")
    .option("header", True)
    .schema(orders_schema)
    .load(source_file)
    .withColumn("ingestion_timestamp", F.current_timestamp())
    .withColumn("source_file", F.lit(source_file))
    .withColumn("batch_id", F.lit(batch_id))
)

orders_bronze_df.printSchema()
display(orders_bronze_df.limit(10))
```

```python
(
    orders_bronze_df.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .saveAsTable("lakehouse.bronze.orders")
)
```

```python
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, DoubleType
from uuid import uuid4
import logging

logger = logging.getLogger("market_data")
logger.setLevel(logging.INFO)

if not logger.handlers:
    handler = logging.StreamHandler()
    handler.setLevel(logging.INFO)
    logger.addHandler(handler)

orders_bronze_schema = StructType([
    StructField("order_id", StringType(), True),
    StructField("customer_id", StringType(), True),
    StructField("order_status", StringType(), True),
    StructField("order_purchase_timestamp", StringType(), True),
    StructField("order_approved_at", StringType(), True),
    StructField("order_delivered_carrier_date", StringType(), True),
    StructField("order_delivered_customer_date", StringType(), True),
    StructField("order_estimated_delivery_date", StringType(), True),
    # Optional column for corrupted row (when using columnNameOfCorruptRecord option)
    StructField("_corrupt_record", StringType(), True)
])

customers_bronze_schema = StructType([
    StructField("customer_id", StringType(), True),
    StructField("customer_unique_id", StringType(), True),
    StructField("customer_zip_code_prefix", StringType(), True),
    StructField("customer_city", StringType(), True),
    StructField("customer_state", StringType(), True),
    StructField("_corrupt_record", StringType(), True)
])

products_bronze_schema = StructType([
    StructField("product_id", StringType(), True),
    StructField("product_category_name", StringType(), True),
    StructField("product_name_lenght", IntegerType(), True),
    StructField("product_description_lenght", IntegerType(), True),
    StructField("product_photos_qty", IntegerType(), True),
    StructField("product_weight_g", DoubleType(), True),
    StructField("product_length_cm", DoubleType(), True),
    StructField("product_height_cm", DoubleType(), True),
    StructField("product_width_cm", DoubleType(), True),
    StructField("_corrupt_record", StringType(), True)
])

sellers_bronze_schema = StructType([
    StructField("seller_id", StringType(), True),
    StructField("seller_zip_code_prefix", StringType(), True),
    StructField("seller_city", StringType(), True),
    StructField("seller_state", StringType(), True),
    StructField("_corrupt_record", StringType(), True)
])

order_reviews_bronze_schema = StructType([
    StructField("review_id", StringType(), True),
    StructField("order_id", StringType(), True),
    StructField("review_score", IntegerType(), True),
    StructField("review_comment_title", StringType(), True),
    StructField("review_comment_message", StringType(), True),
    StructField("review_creation_date", StringType(), True),
    StructField("review_answer_timestamp", StringType(), True),
    StructField("_corrupt_record", StringType(), True)
])

order_payments_bronze_schema = StructType([
    StructField("order_id", StringType(), True),
    StructField("payment_sequential", IntegerType(), True),
    StructField("payment_type", StringType(), True),
    StructField("payment_installments", IntegerType(), True),
    StructField("payment_value", DoubleType(), True),
    StructField("_corrupt_record", StringType(), True)
])

order_items_bronze_schema = StructType([
    StructField("order_id", StringType(), True),
    StructField("order_item_id", IntegerType(), True),
    StructField("product_id", StringType(), True),
    StructField("seller_id", StringType(), True),
    StructField("shipping_limit_date", StringType(), True),
    StructField("price", DoubleType(), True),
    StructField("freight_value", DoubleType(), True),
    StructField("_corrupt_record", StringType(), True)
])

geolocation_bronze_schema = StructType([
    StructField("geolocation_zip_code_prefix", StringType(), True),
    StructField("geolocation_lat", DoubleType(), True),
    StructField("geolocation_lng", DoubleType(), True),
    StructField("geolocation_city", StringType(), True),
    StructField("geolocation_state", StringType(), True),
    StructField("_corrupt_record", StringType(), True)
])


product_category_name_translation_bronze_schema = StructType([
    StructField("product_category_name", StringType(), True),
    StructField("product_category_name_english", StringType(), True),
    StructField("_corrupt_record", StringType(), True)
])

schemas = {
    'orders': orders_bronze_schema,
    'customers': customers_bronze_schema,
    'products': products_bronze_schema,
    'sellers': sellers_bronze_schema,
    'order_reviews': order_reviews_bronze_schema,
    'order_payments': order_payments_bronze_schema,
    'order_items': order_items_bronze_schema,
    'geolocation': geolocation_bronze_schema,
    'product_category_name_translation': product_category_name_translation_bronze_schema,
}
```

```python
schemas = {
    name: StructType([
        StructField(
            field.name,
            StringType(),
            True
        )
        for field in schema.fields
    ])
    for name, schema in schemas.items()
}
```

```python
from uuid import uuid4

source_path = "/Volumes/lakehouse/raw/olist_source"

source_files = {
    'orders': f"{source_path}/olist_orders_dataset.csv",
    'customers': f"{source_path}/olist_customers_dataset.csv",
    'products': f"{source_path}/olist_products_dataset.csv",
    'sellers': f"{source_path}/olist_sellers_dataset.csv",
    'order_reviews': f"{source_path}/olist_order_reviews_dataset.csv",
    'order_payments': f"{source_path}/olist_order_payments_dataset.csv",
    'order_items': f"{source_path}/olist_order_items_dataset.csv",
    'geolocation': f"{source_path}/olist_geolocation_dataset.csv",
    'product_category_name_translation': f"{source_path}/product_category_name_translation.csv",
}

batch_id = str(uuid4())
```

```python
def ingest_data_to_bronze(source_file, bronze_schema, table_name, batch_id):

    bronze_df = (
        spark.read
        .format("csv")
        .option("header", True)
        .option("multiLine", True)  # review_comment_message
        .option("quote", '"')       # comments
        .option("escape", '"')      # comments
        .option("mode", "PERMISSIVE")
        .option("columnNameOfCorruptRecord", "_corrupt_record")
        .schema(bronze_schema)
        .load(source_file)
        .withColumn("ingestion_timestamp", F.current_timestamp())
        .withColumn("source_file", F.lit(source_file.split("/")[-1]))
        .withColumn("batch_id", F.lit(batch_id))
    )

    (
        bronze_df.write
        .format("delta")
        .mode("overwrite")
        .option("overwriteSchema", "true")
        .saveAsTable(f"lakehouse.bronze.{table_name}")
    )
```

```python

for file in source_files:
        logging.info(f"Starting ingestion: {file}, batch_id={batch_id}")
        ingest_data_to_bronze(source_file=source_files[file], bronze_schema=schemas[file], table_name=file, batch_id=batch_id)
        logging.info(f"Completed ingestion: {file}")
```

```python
from pyspark.sql import functions as F

validation_results = []

for table_name in schemas.keys():

    table = f"lakehouse.bronze.{table_name}"
    df = spark.table(table)

    validation_results.append((
        table_name,
        df.count(),
        len(df.columns),
        df.filter(F.col("batch_id").isNull()).count(),
        df.select("batch_id").distinct().count()
    ))

validation_df = spark.createDataFrame(
    validation_results,
    [
        "table_name",
        "row_count",
        "column_count",
        "missing_batch_ids",
        "distinct_batches"
    ]
)

display(validation_df)
```
